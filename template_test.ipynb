```python
df = vitaldb.load_clinical_data(list(range(1, 1000000)))

print("All clinical metadata")
print("shape:", df.shape)
print("columns:", list(df.columns))
print("\nASA columns:", [c for c in df.columns if "asa" in c.lower()])
df.iloc[:5, :20]
```

```python

vc = df["asa"].value_counts(dropna=False).sort_index()
out = pd.DataFrame({
    "count": vc.astype(int),
    "percent": (vc / len(df) * 100).round(2),
})
print("\nASA distribution: all cases")
print(out.to_string())
```

```python
caseids = sorted(map(int, vitaldb.find_cases(["PLETH", "ECG_II"])))
sub = df[df["caseid"].isin(caseids)]

vc2 = sub["asa"].value_counts(dropna=False).sort_index()
out2 = pd.DataFrame({
    "count": vc2.astype(int),
    "percent": (vc2 / len(sub) * 100).round(2),
})
print("\nASA distribution: cases with PLETH + ECG_II")
print("n_cases:", len(sub))
print(out2.to_string())
```

```python





for n in [20, 80, 200, 500, len(caseids)]:
    tmp_ids = caseids[:n]
    tmp = df[df["caseid"].isin(tmp_ids)]
    vc = tmp["asa"].value_counts(dropna=False).sort_index()
    out = pd.DataFrame({
        "count": vc.astype(int),
        "percent": (vc / len(tmp) * 100).round(2),
    })
    print(f"\nASA distribution: first {n} PLETH+ECG_II cases")
    print(out.to_string())

```

```python
caseid = caseids[0]
track_names = vitaldb.get_track_names(caseid)
print(f"\ntracks in first available caseid={caseid}:")
print(track_names[:50])
```

```python





data = vitaldb.load_case(caseid, ["PLETH", "ECG_II"], 1 / 100)
print(f"\nloaded waveform array for caseid={caseid}, tracks=[PLETH, ECG_II], interval=0.01s")
print("shape:", data.shape)
print("first 3 rows:")
print(pd.DataFrame(data, columns=["PLETH", "ECG_II"]).head(3).to_string())
```
